# Прямоугольные импульсы (diff)

Максимумы diff (фронт вверх) → period, от max до min → pulse, duty = pulse/period.

Три захвата, 500 kHz. ТЭС — **CH2**.

```python
import importlib
import numpy as np
import matplotlib.pyplot as plt

import pulse_analysis
importlib.reload(pulse_analysis)
from pulse_analysis import read_sigrok_analog, read_sigrok_samplerate, analyze_rect_pulses

DATASETS = [
    ("etal1khz_05M.zip", 1, 1000),
    ("etal80kHz_05M.zip", 1, 80000),
    ("tes1.5_05M.zip", 2, 77.7),
]

runs = {}
for path, ch, f_exp in DATASETS:
    fs = read_sigrok_samplerate(path)
    sig = read_sigrok_analog(path, channel=ch)
    ps = analyze_rect_pulses(sig, fs)
    s = ps.steady
    runs[path] = dict(fs=fs, sig=sig, ps=ps, f_exp=f_exp, ch=ch)
    print(path, "ch", ch)
    print("  n", ps.period.size, "steady", s.sum())
    print("  T_s", np.median(ps.period[s]))
    print("  f_Hz", 1 / np.median(ps.period[s]))
    print("  duty", np.nanmedian(ps.duty[s]))
    print("  pulse_s", np.nanmedian(ps.pulse[s]))
```

```python
path = "tes1.5_05M.zip"
r = runs[path]
ps, sig, fs = r["ps"], r["sig"], r["fs"]
s = ps.steady
t = np.arange(len(sig)) / fs
T = np.median(ps.period[s])
win = int(5 * T * fs)

fig, ax = plt.subplots(3, 1, figsize=(12, 7), sharex=True)
ax[0].plot(ps.t_center[s], ps.freq[s], ".", ms=2)
ax[0].axhline(1 / T, c="k", ls="--")
ax[0].set_ylabel("freq, Hz")
ax[1].plot(ps.t_center[s], ps.duty[s], ".", ms=2)
ax[1].set_ylabel("duty")
ax[2].plot(t[:win], sig[:win], lw=0.5)
ax[2].set_ylabel("V")
ax[2].set_xlabel("t, s")
for a in ax:
    a.grid(True, alpha=0.3)
plt.suptitle(path)
plt.tight_layout()
plt.show()
```
