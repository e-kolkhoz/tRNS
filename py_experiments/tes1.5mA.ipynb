# ТЭС 1.5 mA

Эталоны и общая логика — `rect_pulses.ipynb`. Здесь только этот захват.

```python
import numpy as np
import matplotlib.pyplot as plt
from pulse_analysis import read_sigrok_analog, read_sigrok_samplerate, analyze_rect_pulses

PATH = "tes1.5_05M.zip"
CH = 2

fs = read_sigrok_samplerate(PATH)
sig = read_sigrok_analog(PATH, channel=CH)
ps = analyze_rect_pulses(sig, fs)
s = ps.steady

T = np.median(ps.period[s])
f = 1 / T
print(PATH, "ch", CH, "fs", fs)
print("n", ps.period.size, "steady", s.sum())
print("T_s", T, "f_Hz", f)
m = s
print("duty", np.nanmedian(ps.duty[m]), "pulse_s", np.nanmedian(ps.pulse[m]))
```

```python
t = np.arange(len(sig)) / fs
win = int(5 * T * fs)
plt.figure(figsize=(12, 3))
plt.plot(t[:win], sig[:win], lw=0.5)
plt.xlabel("t, s")
plt.ylabel("V")
plt.grid(True, alpha=0.3)
plt.show()
```
